```python
import pandas as pd
import numpy as np
df = pd.read_csv('weather.csv')
df
```

```python
print(df.head())
print(df.info())
print(df.describe())
```

```python
# Check null values
print(df.isnull().sum())
# df.dropna(inplace=True) --> this will remove all the rows with null values 
df.fillna(df.mean(numeric_only=True), inplace=True) # fill null values with mean of the numeric column
# inplace makes changes in the original dataset instead of creating a new one
for col in df.select_dtypes(include='object'):
    df[col] = df[col].fillna(df[col].mode()[0])
df = df.drop_duplicates() #removes dulpicates 
df.
```

```python
x = df.drop(['RainTomorrow','RISK_MM'], axis=1)
y = df['RainTomorrow']
x=pd.get_dummies(x)
print(x)
```

```python
from sklearn.model_selection import train_test_split 
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=1)
print(x_train) # model learns patterns from training set
print(x_test) # to check how good that model is
print(y_train) 
print(y_test)  
```

```python
from sklearn.preprocessing import MinMaxScaler
mm= MinMaxScaler()
num = x_train.select_dtypes(include='number').columns
x_train[num] = mm.fit_transform(x_train[num])
x_test[num]=mm.transform(x_test[num])
print(x_train[num])
print(x_test[num])
```
